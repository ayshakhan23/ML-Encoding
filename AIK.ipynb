```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import linear_model
import plotly.express as px
from word2number import w2n
```

```python
df= pd.read_csv("homeprices-town.csv")
df
```

```python
dumiesdf = pd.get_dummies(df.town)
dumiesdf
```

```python
dumiesdf['monroe township']=dumiesdf['monroe township'].replace({False:0,True:1})
dumiesdf['monroe township']
```

```python
dumiesdf = dumiesdf.astype(int)
dumiesdf
```

```python
merge =pd.concat([df,dumiesdf],axis='columns')
merge
```

```python
final = merge.drop(['town','west windsor'],axis='columns')
final
```

```python
y = final.price
```

```python
x= final.drop('price',axis='columns')
x
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
```

```python
model.fit(x,y)
```

```python
model.predict(x)
```

```python
model.score(x,y)
```

```python
model.predict([[5000,1,0]])
```
